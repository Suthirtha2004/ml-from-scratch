Implementing softmax and cross entropy loss in Tensorflow

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import make_blobs
```

```python
data = [[-5,2],[-2,2],[1,2],[5,2]]
X_train , y_train = make_blobs(n_samples = 2000,centers=data,cluster_std=1.0,random_state=30)

# create model

model = Sequential(
    [
        Dense(25, activation='relu'),
        Dense(15,activation='relu'),
        Dense(4,activation='softmax')
    ]
)

model.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy(),
    optimizer = tf.keras.optimizers.Adam(0.001),
)

model.fit(
    X_train,y_train,
    epochs = 10
)
```

```python
results = model.predict(X_train)
print(results[::])
print(len(results))
print("Largest value" , np.max(results) , "Smallest value",np.min(results))
```

Now using more precise calculation we are going to implement logits.The logits function helps us to use the loss function un the softmaz operation in the loss calculation.

```python
# Preferred Model

model2 = Sequential([
    Dense(25,activation= 'relu'),
    Dense(15,activation='relu'),
    Dense(4,activation='linear') # This is used because we will be calculating loss in the loss function
])

model2.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer = tf.keras.optimizers.Adam(0.001)
)

model2.fit(
    X_train,y_train,
    epochs = 10
)
```

The output we get are not probabilities they are just the computed values before the final softmax activation layer hits

```python
unprocessed_results = model2.predict(X_train)
results = tf.nn.softmax(unprocessed_results).numpy()

print(f"Two output vectors:\n {results[:2]}")
print("Largest value",np.max(results),"Smallest value",np.min(results))
```
