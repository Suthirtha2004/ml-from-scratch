```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Make 4-class dataset for classification

classes = 4
m = 100
centers = [[-5,2],[-2,-2],[1,2],[5,-2]]
std = 1.0
X_train , y_train = make_blobs(n_samples=m,centers=centers,cluster_std=std,random_state=30)
```

```python
# Show classes in dataset
print(f"Unique classes {np.unique(y_train)}")
# Show classes are represented
print(f"Class representation {y_train[:10]}")
#Show shapes of our dataset
print(f"Shape of X_train : {X_train.shape} , shape of y_train : {y_train.shape}")
```

2-layer Network model the output layer uses a linear and then a softmax activation.

```python
tf.random.set_seed(1234)

model = Sequential(
    [
        Dense(2,activation='relu',name = "L1"),
        Dense(4,activation ='linear',name = "L2")
    ]
)
```

```python
model.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer = tf.keras.optimizers.Adam(0.01)
)

model.fit(
    X_train,y_train,
    epochs = 200
)
```

```python
pre_results = model.predict(X_train)
results = tf.nn.softmax(pre_results).numpy()

print(f"The output probabilites of top 10 , {results[:10]}")
```
