Here we will be using Numpy to write a simple neural network from scratch

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
import numpy as np

def generate_coffee_data(
    n_samples=200,
    temperature_range=(175, 260),
    duration_range=(8, 18)
):
    rng = np.random.default_rng(42)

    X = rng.random(n_samples * 2).reshape(-1, 2)

    # Temperature
    X[:, 0] = (
        X[:, 0] * (temperature_range[1] - temperature_range[0])
        + temperature_range[0]
    )

    # Roasting duration
    X[:, 1] = (
        X[:, 1] * (duration_range[1] - duration_range[0])
        + duration_range[0]
    )

    Y = np.zeros(len(X))

    for i, (temperature, duration) in enumerate(X):

        # Good roasting requires both sufficient temperature
        # and an appropriate roasting duration.
        if 190 <= temperature <= 240 and 10 <= duration <= 15:
            Y[i] = 1

    return X, Y.reshape(-1, 1)
```

```python
X,Y = generate_coffee_data()
print(X.shape,Y.shape)
```

```python
print(f"Temperature Max and Min values : {np.max(X[:,0]):0.2f}, {np.min(X[:,0]):0.2f}")
print(f"Duration Max and Min values : {np.max(X[:,1]):0.2f}, {np.min(X[:,1]):0.2f}")

norm_layer = tf.keras.layers.Normalization(axis=-1)
norm_layer.adapt(X)
Xn = norm_layer(X)

print(f"Temp after normalization {np.max(Xn[:,0]):0.2f},{np.min(Xn[:,0]):0.2f}")
print(f"Duration after normalization {np.max(Xn[:,1]):0.2f},{np.min(Xn[:,1]):0.2f}")
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
#This is equivalent to Dense
g = sigmoid
def my_dense(a_in,W,b):
  """
    Computes dense layer
    Arguments:
    a_in (ndarray(n,)) : Data
    W(ndarray(n,j)) : n features and j units
    b (ndarray(j,)) : j units
  """
  units = W.shape[1]
  a_out = np.zeros(units)
  for j in range(units):
    w = W[:,j]
    z = np.dot(w,a_in) + b[j]
    a_out[j] = g(z)

  return (a_out)
```

```python
#This is equivalent to Sequential
def my_sequential(x,W1,b1,W2,b2):
  a1 = my_dense(x,W1,b1)
  a2 = my_dense(a1,W2,b2)

  return (a2)
```

```python
W1_tmp = np.array( [[-8.91,  0.30, 12.8 ], [-0.2,  -7.33, 10.83]] )
b1_tmp = np.array( [-9.83, -9.30,  0.97] )
W2_tmp = np.array( [[-32.18], [-26.59], [-33.56]] )
b2_tmp = np.array( [15.42] )
```

```python
# Equivalent of model predict

def my_predict(X,W1,b1,W2,b2):
  m = X.shape[0]
  pre = np.zeros((m,1))
  for i in range(m):
    pre[i,0] = my_sequential(X[i],W1,b1,W2,b2)

  return pre
```

```python
# Testing
X_tst = np.array([
    [200,13.9],  # postive example
    [200,17]])   # negative example
X_tstn = norm_layer(X_tst)
predictions = my_predict(X_tst,W1_tmp,b1_tmp,W2_tmp,b2_tmp)

ycap = np.zeros_like(predictions)
for i in range(len(predictions)) :
  if predictions[i] >= 0.5:
    ycap[i] = 1
  else:
    ycap[i] = 0

print(f"Output =\n {ycap}")
```
