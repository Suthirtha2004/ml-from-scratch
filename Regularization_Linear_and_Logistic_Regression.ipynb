Cost Function with Regularization for Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def compute_cost_linear_reg(x,y,w,b,lambda_ =1 ):
  """
  Computes the cost over all the samples
  Args:

  x(ndarry(m,n)) , y(ndarray(m,)) , w(ndarray(n,))
  b(scalar)
  lambda(scalar) : Controls the regularization

  Returns :
    total_cost(scalar)
  """

  m = x.shape[0]
  n = len(w)

  cost = 0
  for i in range(m):
    fwb_i = np.dot(x[i],w) + b
    cost = cost + (fwb_i - y[i])**2

  cost = cost/(2*m)

  reg_cost = 0
  for j in range(n):
    reg_cost += (w[j]**2)

  reg_cost = (lambda_ /(2*m))*reg_cost

  total_cost = cost + reg_cost
  return total_cost
```

```python
np.random.seed(1)
x_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(x_tmp.shape[1]).reshape(-1,) - 0.5
lambda_tmp = 0.7
b_tmp = 0.5
cost_tmp = compute_cost_linear_reg(x_tmp,y_tmp,w_tmp,b_tmp,lambda_tmp)

print("Regularized cost:" , cost_tmp)
```

Gradient Descent for Linear Regression

```python
def compute_gradient_linear_Reg(x,y,w,b,lambda_):
  """
  Computes the gradient for Linear Regression
  Args:
    x(ndarray(m,n)) y(ndarray(m,)) w(ndarray(n,))
    b(scalar) lambda_ (scalar)

  Returns :
  dj_dw(ndrray(n,)) : The grad of cost w.r.t to weight
  dj_db(scalar) : The grad of cost w.r.t to bias
  """

  m,n = x.shape
  dj_dw = np.zeros((n,))
  dj_db = 0

  for i in range(m):
    err = (np.dot(x[i],w) + b) - y[i]
    for j in range(n):
      dj_dw[j] = dj_dw[j] + err*x[i,j]

    dj_db = dj_db + err

  dj_dw = dj_dw/m
  dj_db = dj_db/m

  for j in range(n):
    dj_dw[j] = dj_dw[j] + (lambda_ /m)*w[j]

  return dj_dw,dj_db
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_dw_tmp, dj_db_tmp =  compute_gradient_linear_Reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print(f"dj_db: {dj_db_tmp}", )
print(f"Regularized dj_dw:\n {dj_dw_tmp.tolist()}")
```

Regularized Cost Function for Logistic Regression

```python
def sigmoid(z):
  g = 1/(1+ np.exp(-z))
  return g
```

```python
def compute_cost_logistic_Reg(x,y,w,b,lambda_ = 1):
  """
  Computes the cost over all the samples
  Args:

  x(ndarry(m,n)) , y(ndarray(m,)) , w(ndarray(n,))
  b(scalar)
  lambda(scalar) : Controls the regularization

  Returns :
    total_cost(scalar)
  """
  m,n = x.shape
  cost = 0
  total_cost = 0
  for i in range(m):
    z_i = np.dot(x[i],w) + b
    fwb_i = sigmoid(z_i)
    cost += -y[i]*np.log(fwb_i) - (1 - y[i])* np.log(1 - fwb_i)

  cost = cost/m

  reg_Cost = 0
  for j in range(n):
    reg_Cost += w[j]**2

  reg_Cost = (lambda_ /(2*m)) * reg_Cost
  total_cost = cost + reg_Cost
  return total_cost
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
cost_tmp = compute_cost_logistic_Reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print("Regularized cost:", cost_tmp)
```

Regularized Gradient Descent for Logistic Regression

```python
def compute_grad_logistic_reg(x,y,w,b,lambda_):
  """
  Computes the gradient for Linear Regression
  Args:
    x(ndarray(m,n)) y(ndarray(m,)) w(ndarray(n,))
    b(scalar) lambda_ (scalar)

  Returns :
  dj_dw(ndrray(n,)) : The grad of cost w.r.t to weight
  dj_db(scalar) : The grad of cost w.r.t to bias
  """
  m,n = x.shape
  dj_dw = np.zeros((n,))
  dj_db = 0.0

  for i in range(m):
    fwb_i = sigmoid(np.dot(x[i],w)+b)
    err_i = fwb_i - y[i]

    for j in range(n):
      dj_dw[j] = dj_dw[j] + err_i * x[i,j]

    dj_db = dj_db + err_i

  dj_dw = dj_dw/m
  dj_db = dj_db/m

  for j in range(n):
    dj_dw[j] = dj_dw[j] + (lambda_/m) * w[j]

  return dj_dw,dj_db
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_dw_tmp, dj_db_tmp =  compute_grad_logistic_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print(f"dj_db: {dj_db_tmp}", )
print(f"Regularized dj_dw:\n {dj_dw_tmp.tolist()}", )
```
